## Clustering

Nous allons désormais chercher à connaitre les centres d'intérêt au regard du DataFrame *photos*. Quel algorithme de clustering proposeriez-vous ? 

#### Utilisation de l'algorithme DBSCAN via scikit-learn

Utiliser DBSCAN et Créer une nouvelle colonne "clusters_step1" dans le dataframe photos pour le label des clusters.

```python
tmp = DBSCAN(eps=0.002, min_samples=20).fit(photos[["latitude", "longitude"]])
photos["clusters_step1"] = np.array(tmp.labels_)
```

Ensemble des clusters donnés par l'algorithme DBSCAN. -1 est le label pour le bruit.

```python
photos["clusters_step1"].unique()
```

Définition des couleurs pour représenter les couleurs. Il n'y a pas spécialement assez de couleurs mais nous nous en contenterons avec un modulo. Le noir correspond au bruit (-1).

```python
COLORS = ['red', 'blue', 'green', 'purple', 'orange', 'darkred',
             'lightred', 'darkblue', 'darkgreen', 'cadetblue', 'pink', 'lightblue', 'lightgreen',
             'gray', 'black']

def getColor(label):
    if label < 0: #bruit
        return "lightgray"
    else:
        return COLORS[label % len(COLORS)]
    
def changeLabels(x, firstLabel):
    return np.array([firstLabel+xi if xi >= 0 else -1 for xi in x])
```

#### Affichage de la carte avec les clusters

```python
map = folium.Map(location=[LAT_CAEN, LON_CAEN], tiles="cartodbpositron", zoom_start=12)

# Un point pour chaque photo
for ind, row in photos.iterrows():
    folium.CircleMarker([row['latitude'], row['longitude']], radius = 1, color=getColor(row["clusters_step1"])).add_to(map)

# Un marker-popup pour chaque cluster à son centre (hormis le bruit)
for label, row in photos.groupby(['clusters_step1'])[["latitude", "longitude"]].mean().iterrows():
    if label > -1:
        popupText = "Cluster n°"+str(label)
        folium.Marker([row['latitude'], row['longitude']], icon=folium.Icon(color=getColor(label)), popup=popupText).add_to(map)
    
#map.save("map.html")
map
```

**Amélioration** : un cluster n°*i* contient un très grand ensemble de photos. Cette région, trop dense, n'a pas permis à DBSCAN avec les paramètres donnés précédemment de discriminer d'avantage les centres d'intérêt. Il serait intéréssant d'appliquer de nouveau l'algorithme avec des paramètres plus fins pour ce cluster.

*Remarque* : Pour ne pas entrer en collision avec la numérotation de l'étape 1 lors de la redéfinition des clusters sur le cluster *i* pour ce nouveau DBSCAN, créer une nouvelle colonne "clusters_step2" initialisée avec les valeurs "clusters_step1", puis ensuite modifier le numéro de cluster en "clusters_step2" seulement pour les photos du cluster *i*, en utilisant la fonction *changeLabels*

```python
def changeLabels(x, firstLabel):
    return np.array([firstLabel+xi if xi >= 0 else -1 for xi in x])
```

```python
masque_cluster0 = photos["clusters_step1"] == 0
tmp = DBSCAN(eps=0.00055, min_samples=10).fit(photos.loc[masque_cluster0][["latitude", "longitude"]])
photos["clusters_step2"] = photos["clusters_step1"]
photos.loc[masque_cluster0, "clusters_step2"] = changeLabels(tmp.labels_, np.max(photos["clusters_step1"])+1)
```

Nouvel ensemble des clusters. -1 et -100 sont le bruit.

```python
photos["clusters_step2"].unique()
```

#### Affichage de la carte avec les clusters

```python
map = folium.Map(location=[LAT_CAEN, LON_CAEN], tiles="cartodbpositron", zoom_start=12)

# Un point pour chaque photo
for ind, row in photos.iterrows():
    folium.CircleMarker([row['latitude'], row['longitude']], radius = 1, color=getColor(row["clusters_step2"])).add_to(map)

# Un marker-popup pour chaque cluster à son centre (hormis le bruit)
for label, row in photos.groupby(['clusters_step2'])[["latitude", "longitude"]].mean().iterrows():
    if label > -1:
        popupText = "Cluster n°"+str(label)
        folium.Marker([row['latitude'], row['longitude']], icon=folium.Icon(color=getColor(label)), popup=popupText).add_to(map)
    
#map.save("map.html")
map
```

**Nouvelle amélioration** : Nous pouvons encore affiner le découpage en appliquant de nouveau DBSCAN pour le centre historique de Caen (cluster n°*j*), zone encore une fois plus dense. Appliquer la même procédure. 

Créer une colonne "cluster_step3", appliquer la même démarche et afficher les résultats avec les popups.

```python
masque_cluster13 = photos["clusters_step2"] == 13
tmp = DBSCAN(eps=0.00036, min_samples=15).fit(photos.loc[masque_cluster13][["latitude", "longitude"]])
photos["clusters_step3"] = photos["clusters_step2"]
photos.loc[masque_cluster13, "clusters_step3"] = changeLabels(tmp.labels_, np.max(photos["clusters_step2"])+1)
```
